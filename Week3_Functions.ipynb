## **What is a function?**
A function is a block of organized, reusable code that can make your scripts more effective, easier to read, and simple to manage. You can think of functions as little self-contained programs that perform a specific task, which you can use repeatedly in your code. One of the basic principles in good programming is “do not repeat yourself”. In other words, you should avoid having duplicate lines of code in your scripts. Functions are a good way to avoid such situations, and they can save you a lot of time and effort as you don’t need to tell the computer repeatedly what to do every time it does a common task.

You have already used many **built-in** functions, such as `print()`, `len()`, `type()` and `np.mean()`.

```python
def greet():
    print('Printing from Inside of a function!')

# Main Program
greet()

print('Outside from a function')
```

### **Function with Argument**

```python
# function definition with return statement
def find_square(num):
    result = num * num
    return result

# function call
square = find_square(4)

print('Square:', square)
```

```python
# function with two arguments
def add_numbers(num1, num2):
    total = num1 + num2      # 'total', not 'sum': sum() is already a built-in Python function
    print("Sum: ", total)

# function call with two values
add_numbers(2,5)
```

### **Printing vs returning a result**

**Example 1: a function that PRINTS the result**

Predict the output before running.

```python
def add_print(a, b):
    print(a + b)

x = add_print(2, 5)
print("x is:", x)
```

The function **shows** 7 on the screen, but it does not **give it back**. So `x` gets nothing (`None`).

**Example 2: a function that RETURNS the result**

Predict the output before running.

```python
def add_return(a, b):
    return a + b

y = add_return(2, 5)
print("y is:", y)
```

The function **gives back** 7, so `y` stores it. We can now use `y` again, for example `y * 2` gives 14.

**Rule of thumb:** if you want to use the result later (in a calculation, a chart or another function), **return** it. In data analysis, almost every function you write should return its result.

### **Documenting a function: docstrings**

A **docstring** is a short description written in triple quotes as the first line inside a function. It explains what the function does, what it needs, and what it returns. Anyone using your function (including you, a few weeks later) can read it with `help()`.

```python
def find_cube(num):
    """Return the cube of a number.

    num: the number to be cubed
    """
    return num * num * num

print(find_cube(3))
help(find_cube)
```

### **Function with Early exit**

*Note: this example uses an `if` statement. We will cover `if` properly in Week 4. For now, read it as: "if the length or width is negative, stop and return a message".*

```python
def calculate_area(length, width):
    if length < 0 or width < 0:
        return "Invalid input: Length and width cannot be negative."  # Early exit
    area = length * width
    return area

# Valid input
result1 = calculate_area(10, 10)
print("Area:", result1)  # Output: Area: 100
```

```python
# Invalid input
result2 = calculate_area(5, -10)
print("Area:", result2)  # Output: Invalid input: Length and width cannot be negative.
```

**Think about it:** `calculate_area()` returns a **number** for valid input but **text** for invalid input. What would happen if you tried `result2 * 2`? Returning different types from the same function can cause errors later in your code, so try to avoid it in your own functions.

### **Python Function With Arbitrary Arguments**
Sometimes, we do not know in advance the number of arguments that will be passed into a function. To handle this kind of situation, we can use arbitrary arguments in Python.

Arbitrary arguments allow us to pass a varying number of values during a function call. Python collects all the values into a **tuple**.

*Note: this example uses a `for` loop. We will cover loops properly in Week 5. For now, read it as: "go through each number and add it to the result".*

```python
# program to find sum of multiple numbers

def find_sum(*numbers):
    print("numbers is:", numbers)    # shows the tuple Python created
    print("Sum = ", sum(numbers))

# function call with 7 arguments
find_sum(2, 2, 3, 1, 8, 9, 10)

# function call with 2 arguments
find_sum(4, 9)
```

```python
def describe(*numbers):
    return sum(numbers), len(numbers), max(numbers)

total, count, biggest = describe(2, 2, 3, 1, 8, 9, 10)
print("Total:", total, " Count:", count, " Biggest:", biggest)
```

## **Python Local Variables**
When we declare variables inside a function, these variables will have a local scope (within the function). We cannot access them outside the function. These types of variables are called local variables

```python
def greet():

    # local variable
    message = 'Hello'

    print('Local', message)

greet()

# try to access message variable
# outside greet() function
print(message)
```

## **Python Global Variables**
In Python, a variable declared outside of the function or in global scope is known as a global variable. This means that a global variable can be accessed inside or outside of the function.

```python
# declare global variable
message = 'Hello'

def greet():
    # declare local variable
    print('Local', message)

greet()
print('Global', message)
```

**Access and Modify Python Global Variable**

```python
c = 1 # global variable

def printing():
    print(c)

printing()

# Output: 1
```

Here, we can see that we have accessed a global variable from inside a function. However, if we try to modify the global variable from inside a function:

```python
# global variable
c = 1

def add():

     # increment c by 2
    c = c + 2

    print(c)

add()
```

This is because we can only access (read) the global variable, but cannot modify it from inside the function. One way around this is to use the `global` keyword.

```python
# global variable
c = 1

def add():

    # use of global keyword
    global c

    # increment c by 2
    c = c + 2

    print(c)

add()
```

### **Rules of global Keyword**
The basic rules for the global keyword in Python are:

1. When we create a variable inside a function, it is local by default.
2. When we define a variable outside of a function, it is global by default. You don't have to use the global keyword.
3. We use the global keyword to **modify** a global variable inside a function. (Reading a global variable does **not** need the global keyword, as the `printing()` example showed.)
4. Use of the global keyword outside a function has no effect.

### **Good practice: pass values in and return them**

You should be able to **recognise** `global`, but in your own code it is better to avoid it. Functions that change global variables are hard to test and can cause surprising bugs, because any part of the program might change the value.

A cleaner way is to **pass the value in** as an argument and **return** the new value:

```python
def add_two(value):
    return value + 2

c = 1
c = add_two(c)
print(c)
```

# **Functions for Data Analysis**

So far our functions have worked on single numbers. In data analysis, functions usually work on **collections of data**: lists, NumPy arrays and pandas DataFrames.

### **1. Small functions working together**

Good programs are built from **small functions that each do one job**. Below, one function converts marks to percentages, another summarises them, and a third uses both.

Notice that `summarise()` returns **three values**. Python packs them into a **tuple** (Week 2).

```python
import numpy as np

def to_percent(marks, max_mark):
    """Convert raw marks into percentages."""
    return marks / max_mark * 100

def summarise(values):
    """Return the mean, highest and lowest value."""
    return values.mean(), values.max(), values.min()

def marks_report(marks, max_mark):
    """Convert marks to percentages and print a short summary."""
    percents = to_percent(marks, max_mark)
    mean, highest, lowest = summarise(percents)
    print(f"Average: {mean:.1f}%   Highest: {highest:.1f}%   Lowest: {lowest:.1f}%")

raw_marks = np.array([32, 45, 28, 50, 39])
marks_report(raw_marks, max_mark=50)
```

Because each function does one job, we can **reuse** them.

### **3. Write your own, or use a library?**

Let's write our own function to calculate an average, and compare it with NumPy's `np.mean()`.

```python
def average(numbers):
    """Return the average of a list of numbers."""
    return sum(numbers) / len(numbers)

print(average([10, 20, 30]))
print(np.mean([10, 20, 30]))
```
